W5A20 Viết hàm detect_number_type(s) trả về 'int' nếu s là số nguyên, 'float' nếu là số thực, ngược lại 'neither'.

```python
def kiemtra(s):
    if len(s) == 0:
        return 'neither'
    i = 0
    if s[0] == '-': 
         if len(s) == 1:
            return 'neither'
        i = 1
    dot_count = 0
    digit_count = 0

    for j in range(i, len(s)):
        if s[j] == '.':
            dot_count += 1
        elif '0' <= s[j] <= '9':
            digit_count += 1
        else:
            return 'neither' 
        
    if dot_count == 0 and digit_count > 0:
        return 'int'
    if dot_count == 1 and digit_count > 0:
        return 'float'

    return 'neither'

n = input()
print(kiemtra(n))
```

```python
def sum_halves(s):
    dau = 0
    cuoi = 0
    for i in range(3):
        dau = dau * 10 + (ord(s[i]) - ord('0'))
    for i in range(3, 6):
        cuoi = cuoi * 10 + (ord(s[i]) - ord('0'))

    return dau + cuoi

n = input()
print(sum_halves(n)) 
```

W5A27 Viết hàm trả về số lượng chữ số khác 0 trong n số nguyên nhập vào từ bàn phím (một list)

```python
def  chusokhackhong(lst):
    count = 0
    for num in lst:
        num = abs(num) 
        if num == 0:
            continue
        while num > 0:
            digit = num % 10
            if digit != 0:
                count += 1
            num //= 10
    return count

n = input()
print(chusokhackhong(n))
```

W5A28 Viết hàm tính góc ∠ABC (độ) từ tọa độ A(x1,y1), B(x2,y2), C(x3,y3)

```python
def angle_ABC(x1, y1, x2, y2, x3, y3):
    BAx = x1 - x2
    BAy = y1 - y2
    BCx = x3 - x2
    BCy = y3 - y2

    dot = BAx * BCx + BAy * BCy
    lenBA = (BAx**2 + BAy**2) ** 0.5
    lenBC = (BCx**2 + BCy**2) ** 0.5

    cos_val = dot / (lenBA * lenBC)

    # tránh lỗi do làm tròn
    if cos_val > 1: 
        cos_val = 1
    if cos_val < -1: 
        cos_val = -1
    angle = 180 * (1 - cos_val) / 2  

    return angle

# test
x1, y1 = map(float, input().split())
x2, y2 = map(float, input().split())
x3, y3 = map(float, input().split())
print(angle_ABC(x1,y1,x2,y2,x3,y3))
```

W5A30 Viết hàm tính hệ số nhị thức C(n, k)

```python
def heso(n, k):
    res = 1
    for i in range(1, k+1):
        res = res * (n - i + 1) // i
    return res

# test
n, k = map(int, input().split())
print(heso(n, k))
```

```python
def smart_taxi_fare(car_type, start_hour, km, wait):
    if car_type == 'A':
        fare = 10000
        per_km = 12000
    else:
        fare = 15000
        per_km = 15000

    fare += km * per_km
    fare += wait * 500

    if start_hour >= 22 or start_hour < 5:
        fare *= 1.2

    if km > 10:
        fare *= 0.9

    return int(fare)
car = input()
hour = int(input())
km = float(input())
wait = int(input())

print(smart_taxi_fare(car, hour, km, wait))
```
