Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu mỗi phần tử x trong mảng xuất hiện đúng x lần.
Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp.


```python
n = int(input())
a = list(map(int, input().split()))
dem = {}
xoa = 0
for i in a:
    if i not in dem:
        dem[i] = 1
    else:
        dem[i] = dem[i] + 1
for so in dem:
    solan = dem[so]
    if solan == so:
        continue
    elif solan > so:
       xoa = xoa + (solan - so)
    else:
        xoa = xoa + solan
print(xoa)
```

Tìm phần tử xuất hiện nhiều hơn n/2 lần trong mảng (nếu có) với độ
phức tạp thời gian là O(n), độ phức tạp bộ nhớ là O(1), - Boyer–Moore. 

```python
n = int(input())
a = list(map(int, input().split()))
dem = 0
dem2 = 0
x = 0
for i in a:

    if dem == 0:
        x = i
        dem = 1
    elif i == x:
        dem = dem + 1
    else:
        dem = dem - 1
for i in a:
    if i == x:
        dem2 = dem2 + 1
if dem2 > n // 2:
    print(x)
else:
    print("Khong co")
```

W7A22
Cho N tên sinh viên gồm cả họ và tên, tìm xem tên nào hay gặp nhất.
Nếu có nhiều tên cùng số lần gặp thì lấy tên có thứ tự từ điển nhỏ nhất

```python
n = int(input())
dem = {}
for i in range(n):
    s = input().strip()

    ten = s.split()[-1]

    if ten not in dem:
        dem[ten] = 1
    else:
        dem[ten] = dem[ten] + 1

max_dem = 0
kq = ""

for ten in dem:

    if dem[ten] > max_dem:
        max_dem = dem[ten]
        kq = ten

    elif dem[ten] == max_dem:

        if ten < kq:
            kq = ten

print(kq)
```

W7A25
Tìm số có đúng k chữ số lớn nhất có thể trong một số có n chữ số (coi đầu vào là một số rất lớn - str). Không được hoán đổi vị trí các chữ số


```python
s = input()

k = int(input())

xoa = len(s) - k

stack = []

for so in s:

    while xoa > 0 and len(stack) > 0 and stack[-1] < so:
        stack.pop()
        xoa = xoa - 1

    stack.append(so)

while xoa > 0:
    stack.pop()
    xoa = xoa - 1

for i in range(k):
    print(stack[i], end="")
```

W7A27
Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra xem dãy ngoặc đã cho có hợp lệ
hay không

```python
s = input()

stack = []

hop_le = True

for i in s:

    if i == "(" or i == "[" or i == "{":
        stack.append(i)

    else:

        if len(stack) == 0:
            hop_le = False
            break

        top = stack[-1]

        if i == ")" and top == "(":
            stack.pop()

        elif i == "]" and top == "[":
            stack.pop()

        elif i == "}" and top == "{":
            stack.pop()

        else:
            hop_le = False
            break

if len(stack) != 0:
    hop_le = False

if hop_le:
    print("Hop le")
else:
    print("Khong hop le")
```

nào <= N (N là số nguyên dương cho trước). Nếu có nhiều dãy cùng độ
dài thì lấy dãy nhỏ nhất (gặp đầu tiên từ trái qua)

```python
n = int(input())

a = []
i = n + 1

while len(a) < 10:

    uoc = 0

    for j in range(1, i + 1):
        if i % j == 0:
            uoc = uoc + 1

    if uoc > 2:
        a.append(i)

    else:
        break

    i = i + 1

for i in a:
    print(i, end=" ")
```

W7A29
Tìm Dãy con tăng dần liên tục dài nhất của một danh sách nhập vào từ
bà

```python
n = int(input())

a = list(map(int, input().split()))

dai = 1
max_dai = 1

vt = 0
start = 0

for i in range(1, n):

    if a[i] > a[i - 1]:
        dai = dai + 1

    else:
        dai = 1
        start = i

    if dai > max_dai:
        max_dai = dai
        vt = start

for i in range(vt, vt + max_dai):
    print(a[i], end=" ")
```

W7A30
Tìm Dãy con tăng dần KHÔNG liên tục dài nhất của một danh sách nhập
vào từ bàn phím 

```python
n = int(input())

a = list(map(int, input().split()))

dp = [1] * n

for i in range(n):

    for j in range(i):

        if a[j] < a[i]:

            if dp[j] + 1 > dp[i]:
                dp[i] = dp[j] + 1

print(max(dp))
```
